```python
import pandas as pd
from scipy.io import arff

file_path = 'KDDTest+.arff'

# Загружаем ARFF-файл
# Функция loadarff возвращает кортеж: (данные, метаданные)
data, meta = arff.loadarff(file_path)

```

```python
df = pd.DataFrame(data, columns=meta.names())
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
encoder.fit(df[["protocol_type"]])
X_train_encoded = encoder.transform(df[["protocol_type"]])

```

```python
df["normal_class"] = (df["class"] != b"normal").astype(int)   # 0 = норма, 1 = атака
print(df["normal_class"].value_counts())
print("Типов атак в class:", df["class"].nunique() - 1)
```

```python
x = df.drop(columns=["class", "normal_class", 'X'])
y = df["normal_class"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y,random_state=42)
```
